```python
import pandas as pd

pd.Series(["Apples", "Bananas", "Pineapples"])
```

```python
pd.DataFrame(
    [
        ["Apples", 40, 100],
        ["Bananas", 30, 200],
        ["Pineapples", 70, 50]
    ],
    columns = ["Name", "Price", "Quantity"],
    index = ["Apples", "Bananas", "Pineapples"],
)
```

```python
pd.DataFrame(
    {
        "Name": ["Apples", "Bananas", "Pineapples"],
        "Price": [40, 50, 100],
        "Quantity": [10, 20, 50],
    }
)
```
